```python
import numpy as np
import pandas as pd
from scipy.spatial.distance import cdist
import matplotlib.pyplot as plt
```

```python
# Dataset: 4 instances with 2 numeric attributes

data = np.array([
    [1, 2],
    [2, 3],
    [5, 5],
    [6, 4]
])

print("Dataset:")
print(data)
```

```python
# Number of instances
n = len(data)

# Initialize n x n zero matrix
D = np.zeros((n, n))

# Calculate pairwise Euclidean distances manually
for i in range(n):
    for j in range(i + 1, n):

        distance = np.sqrt(
            (data[i][0] - data[j][0]) ** 2 +
            (data[i][1] - data[j][1]) ** 2
        )

        D[i][j] = distance
        D[j][i] = distance

print("Dissimilarity Matrix (Manual):")
print(D)
```

```python
labels = ["P1", "P2", "P3", "P4"]

D_df = pd.DataFrame(
    D,
    index=labels,
    columns=labels
)

print("Dissimilarity Matrix:")
print(D_df.round(3))
```

```python
# Calculate Euclidean distance using scipy

D_library = cdist(
    data,
    data,
    metric="euclidean"
)

print("Dissimilarity Matrix (Using SciPy):")
print(D_library.round(3))
```

```python
  if np.allclose(D, D_library):
    print("Both matrices are equal.")
else:
    print("Matrices are different.")
```

```python
plt.figure(figsize=(6, 5))

plt.imshow(D, cmap="viridis")

plt.colorbar(label="Euclidean Distance")

plt.xticks(range(len(labels)), labels)
plt.yticks(range(len(labels)), labels)

plt.title("Dissimilarity Matrix Heatmap")
plt.xlabel("Instances")
plt.ylabel("Instances")

plt.show()
```
